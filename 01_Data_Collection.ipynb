```python
import soccerdata as sd
import pandas as pd
```

```python
# Initialize the scraper & getting the Laliga data for 4 seasons - 21/22, 22/23, 23/24, 24/25
# The soccerdata package scrapes data from FBref.com (Football Reference)

fb = sd.FBref(leagues=["ESP-La Liga"], seasons=["2122", "2223", "2324", "2425"])  # SPA = La Liga
```

```python
# Load the fixtures data

fixtures = fb.read_team_match_stats(stat_type="schedule")
```

```python
# Load the shooting data

shooting = fb.read_team_match_stats(stat_type="shooting")
```

```python
# Load the passing, goal creating actions and possession data

passing = fb.read_team_match_stats(stat_type="passing")

gsc = fb.read_team_match_stats(stat_type="goal_shot_creation")

possession = fb.read_team_match_stats(stat_type="possession")
```

```python
# Reset index so multi-index columns (league, game, team & season) become normal columns

for df in [shooting, passing, gsc, possession, fixtures]:
    df.reset_index(inplace=True)
```

```python
# Export each dataframe as CSV
fixtures.to_csv('fixtures_raw.csv', index=False)
shooting.to_csv('shooting_raw.csv', index=False)
passing.to_csv('passing_raw.csv', index=False)
possession.to_csv('possession_raw.csv', index=False)
gsc.to_csv('gsc_raw.csv', index=False)
```
